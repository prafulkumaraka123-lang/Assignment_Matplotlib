## Q1. What is Matplotlib? Explain `plt.plot()` vs `plt.scatter()`.

**Answer:** Matplotlib is a Python library used for data visualization. `plt.plot()` is mainly used for line plots and shows trends by connecting points. `plt.scatter()` displays individual observations as points and is useful for studying relationships between numerical variables.


```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 7, 6]

plt.plot(x, y, marker="o")
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()

plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.show()
```

## Q2. Figure, Axes, `subplot()` and `subplots()`.

**Answer:** A Figure is the overall Matplotlib container. An Axes is an individual plotting area inside a Figure. `plt.subplot()` creates/selects one subplot at a time, while `plt.subplots()` creates a Figure and one or more Axes objects together.


```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [1, 4, 9, 16]

plt.figure(figsize=(8, 3))
plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title("subplot()")

plt.subplot(1, 2, 2)
plt.scatter(x, y)
plt.title("subplot()")
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(8, 3))
axes[0].plot(x, y)
axes[0].set_title("Axes 1")
axes[1].scatter(x, y)
axes[1].set_title("Axes 2")
plt.tight_layout()
plt.show()
```

## Q3. Customize a sine-wave line plot.

**Answer:** `color` controls color, `linestyle` controls line style, `linewidth` controls thickness, `marker` marks data points, and `label` names the line for the legend.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

plt.plot(x, y, color="blue", linestyle="--", linewidth=2,
         marker="o", markersize=3, markevery=10, label="sin(x)")
plt.title("Customized Sine Wave")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.legend()
plt.grid(True)
plt.show()
```

## Q4. What is Seaborn? Explain `hue`.

**Answer:** Seaborn is a statistical visualization library built on Matplotlib. It provides a higher-level interface and works well with Pandas. `hue` maps a categorical variable to color, allowing different groups to be distinguished visually.


```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="day")
plt.title("Seaborn Scatter Plot with hue")
plt.show()
```

## Q5. `plt.bar()` vs `plt.barh()`.

**Answer:** `plt.bar()` creates vertical bars; `plt.barh()` creates horizontal bars.


```python
import matplotlib.pyplot as plt

categories = ["A", "B", "C", "D"]
values = [23, 17, 35, 29]

plt.bar(categories, values)
plt.title("Vertical Bar Chart")
plt.xlabel("Category")
plt.ylabel("Value")
plt.show()

plt.barh(categories, values)
plt.title("Horizontal Bar Chart")
plt.xlabel("Value")
plt.ylabel("Category")
plt.show()
```

## Q6. Histogram.

**Answer:** A histogram displays the distribution of numerical data by dividing values into bins and showing frequencies. Its shape can show center, spread, skewness, multiple peaks, and unusual values.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
data = np.random.randn(1000)

plt.hist(data, bins=30)
plt.title("Histogram of 1000 Random Values")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## Q7. Pie chart.

**Answer:** `autopct` displays percentages, `startangle=120` rotates the starting angle, and `plt.axis("equal")` keeps the pie circular instead of stretched.


```python
import matplotlib.pyplot as plt

sizes = [40, 25, 20, 15]
labels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.pie(sizes, labels=labels, autopct="%1.1f%%", startangle=120)
plt.title("Fruit Percentage Share")
plt.axis("equal")
plt.show()
```

## Q8. `plt.annotate()`.

**Answer:** `plt.annotate()` adds explanatory text to a plot and can point to a location with an arrow. It is useful for highlighting important points such as a peak.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 200)
y = np.sin(x)

plt.plot(x, y, label="sin(x)")
plt.annotate("Peak",
             xy=(np.pi/2, 1),
             xytext=(np.pi/2 + 0.6, 1.2),
             arrowprops=dict(arrowstyle="->"))
plt.title("Sine Wave with Peak Annotation")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.legend()
plt.grid(True)
plt.show()
```

## Q9. Seaborn histogram and boxplot.

**Answer:** The scores are placed in a DataFrame with column `Score`. `sns.histplot()` shows the distribution and `sns.boxplot()` shows median, quartiles, whiskers and possible outliers. For the given data, the standard 1.5×IQR rule does not identify any numerical outliers.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

scores = [45, 67, 89, 23, 56, 78, 90, 34]
df = pd.DataFrame({"Score": scores})

print(df)

sns.histplot(data=df, x="Score", bins=5)
plt.title("Histogram of Scores")
plt.show()

sns.boxplot(data=df, x="Score")
plt.title("Boxplot of Scores")
plt.show()

Q1 = df["Score"].quantile(0.25)
Q3 = df["Score"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = df[(df["Score"] < lower) | (df["Score"] > upper)]
print("Potential outliers:")
print(outliers)
```

## Q10. Plotly.

**Answer:** Plotly is a Python visualization library focused on interactive charts. Matplotlib and Seaborn are commonly used for static/statistical visualization, while Plotly allows interaction such as hovering, zooming and panning.


```python
import plotly.express as px
import pandas as pd

x = [1, 2, 3, 4, 5]
y = [10, 14, 22, 18, 27]

df = pd.DataFrame({"x": x, "y": y})

fig = px.scatter(df, x="x", y="y", title="Interactive Scatter Plot")
fig.show()
```
